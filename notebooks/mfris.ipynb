## Simple sin demo

A 1 Hz sine wave completes one cycle per second. NumPy takes angles in radians, so use `np.sin(2 * np.pi * frequency * t)`, with frequency in Hz and time `t` in seconds. The example below samples at 100 samples per second for 2 seconds, producing two complete cycles.

```python
def sin(frequency, duration, sample_rate):
    """Generate a sine wave: frequency in Hz, duration in seconds, sample_rate in Hz."""
    t = np.arange(0, duration, 1 / sample_rate)
    return t, np.sin(2 * np.pi * frequency * t)
```

```python
x, y = sin(frequency=1, duration=2, sample_rate=100)

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, y)
ax.set(xlabel="Time (seconds)", ylabel="Amplitude", title="A 1 Hz sine wave")
fig.tight_layout()
fig.savefig(figures / "sine_wave.png", dpi=150)
plt.show()
```
